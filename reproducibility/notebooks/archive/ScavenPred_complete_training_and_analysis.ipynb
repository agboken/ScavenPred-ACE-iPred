## Exact SHAP recomputation code
```python
import shap
explainer = shap.TreeExplainer(rf)
sv = explainer.shap_values(X)
arr = np.asarray(sv)
sv_pos = np.asarray(sv[1]) if isinstance(sv, list) else (arr[:,:,1] if arr.ndim == 3 else arr)
importance = pd.DataFrame({'Feature': FEATURES, 'Mean_abs_SHAP': np.abs(sv_pos).mean(axis=0)}).sort_values('Mean_abs_SHAP', ascending=False)
shap.summary_plot(sv_pos, X, feature_names=FEATURES)
```
